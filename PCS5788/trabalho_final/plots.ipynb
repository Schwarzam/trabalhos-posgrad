```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch


def box(
    ax,
    x,
    y,
    text,
    width=2.4,
    height=0.75,
    fontsize=11
):
    patch = FancyBboxPatch(
        (x - width / 2, y - height / 2),
        width,
        height,
        boxstyle="round,pad=0.08",
        linewidth=1.5,
        fill=False
    )

    ax.add_patch(patch)

    ax.text(
        x,
        y,
        text,
        ha="center",
        va="center",
        fontsize=fontsize
    )


def arrow(
    ax,
    x1,
    y1,
    x2,
    y2
):
    ax.annotate(
        "",
        xy=(x2, y2),
        xytext=(x1, y1),
        arrowprops=dict(
            arrowstyle="->",
            linewidth=1.5
        )
    )
```

```python
fig, ax = plt.subplots(
    figsize=(7, 10)
)

ax.axis("off")

x = 5

box(
    ax, x, 9,
    "Imagem MNIST\n1 × 28 × 28"
)

box(
    ax, x, 8,
    "Bloco convolucional\nCNN"
)

box(
    ax, x, 7,
    "Feature map\n32 × 7 × 7"
)

box(
    ax, x, 6,
    "Flatten\n1568 features"
)

box(
    ax, x, 5,
    "Fully Connected\n1568 → 20"
)

for y1, y2 in [
    (8.6, 8.4),
    (7.6, 7.4),
    (6.6, 6.4),
    (5.6, 5.4)
]:
    arrow(
        ax,
        x, y1,
        x, y2
    )


positions = [
    1.7,
    3.9,
    6.1,
    8.3
]

for i, px in enumerate(positions):

    box(
        ax,
        px,
        3.6,
        f"PQC {i+1}\n5 qubits",
        width=1.8,
        height=0.9,
        fontsize=10
    )

    arrow(
        ax,
        x,
        4.6,
        px,
        4.08
    )


box(
    ax,
    x,
    2.3,
    "Concatenação\n20 expectation values"
)

for px in positions:

    arrow(
        ax,
        px,
        3.12,
        x,
        2.72
    )


box(
    ax,
    x,
    1.2,
    "Fully Connected\n20 → 10"
)

arrow(
    ax,
    x, 1.9,
    x, 1.6
)


box(
    ax,
    x,
    0.2,
    "10 classes"
)

arrow(
    ax,
    x, 0.8,
    x, 0.6
)


ax.set_xlim(
    0,
    10
)

ax.set_ylim(
    -0.4,
    9.7
)

ax.set_title(
    "HQNN-Parallel",
    fontsize=17,
    fontweight="bold",
    pad=15
)

plt.tight_layout()

plt.show()
```
