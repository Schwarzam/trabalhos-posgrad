```python
import tensorflow as tf

# parâmetros
w1 = tf.Variable(-0.2, dtype=tf.float32)
w2 = tf.Variable( 0.5, dtype=tf.float32)
w3 = tf.Variable( 0.9, dtype=tf.float32)
w4 = tf.Variable(-0.6, dtype=tf.float32)
w5 = tf.Variable( 0.2, dtype=tf.float32)
w6 = tf.Variable(-0.4, dtype=tf.float32)
b1 = tf.Variable( 0.4, dtype=tf.float32)
b2 = tf.Variable(-0.2, dtype=tf.float32)
b3 = tf.Variable(-0.5, dtype=tf.float32)

# dado
x1 = tf.constant(0.6, dtype=tf.float32)
x2 = tf.constant(-0.3, dtype=tf.float32)
y  = tf.constant(1.0, dtype=tf.float32)

with tf.GradientTape() as tape:
    z1 = w1*x1 + w2*x2 + b1
    p1 = tf.sigmoid(z1)

    z2 = w3*x1 + w4*x2 + b2
    p2 = tf.sigmoid(z2)

    z3 = w5*p1 + w6*p2 + b3
    p3 = tf.sigmoid(z3)

    c = (p3 - y)**2

grads = tape.gradient(c, [w1, w2, w3, w4, w5, w6, b1, b2, b3])

nomes = ["dc/dw1", "dc/dw2", "dc/dw3", "dc/dw4", "dc/dw5", "dc/dw6", "dc/db1", "dc/db2", "dc/db3"]

print("Solução item (a):")
print("c =", c.numpy())
for nome, g in zip(nomes, grads):
    print(nome, "=", g.numpy())
```

```python
import tensorflow as tf

# parâmetros
w1 = tf.Variable(-0.2, dtype=tf.float32)
w2 = tf.Variable( 0.5, dtype=tf.float32)
w3 = tf.Variable( 0.9, dtype=tf.float32)
w4 = tf.Variable(-0.6, dtype=tf.float32)
w5 = tf.Variable( 0.2, dtype=tf.float32)
w6 = tf.Variable(-0.4, dtype=tf.float32)
b1 = tf.Variable( 0.4, dtype=tf.float32)
b2 = tf.Variable(-0.2, dtype=tf.float32)
b3 = tf.Variable(-0.5, dtype=tf.float32)

# lote com 2 dados
x1 = tf.constant([0.8, -0.1], dtype=tf.float32)
x2 = tf.constant([-0.1, 0.9], dtype=tf.float32)
y  = tf.constant([1.0, 0.0], dtype=tf.float32)

with tf.GradientTape() as tape:
    z1 = w1*x1 + w2*x2 + b1
    p1 = tf.sigmoid(z1)

    z2 = w3*x1 + w4*x2 + b2
    p2 = tf.sigmoid(z2)

    z3 = w5*p1 + w6*p2 + b3
    p3 = tf.sigmoid(z3)

    # custo médio do lote
    c = tf.reduce_mean((p3 - y)**2)

grads = tape.gradient(c, [w1, w2, w3, w4, w5, w6, b1, b2, b3])

nomes = ["dc/dw1", "dc/dw2", "dc/dw3", "dc/dw4", "dc/dw5", "dc/dw6", "dc/db1", "dc/db2", "dc/db3"]

print("Solução item (b):")
print("c =", c.numpy())
for nome, g in zip(nomes, grads):
    print(nome, "=", g.numpy())
```
