```python
import os; os.environ['TF_CPP_MIN_LOG_LEVEL']='3'
os.environ['TF_FORCE_GPU_ALLOW_GROWTH'] = 'true'

import tensorflow.keras as keras
from tensorflow.keras.models import Model, load_model
from tensorflow.keras.layers import Dense, Conv2D, BatchNormalization, Activation
from tensorflow.keras.layers import GlobalAveragePooling2D, Input, Flatten
from tensorflow.keras.regularizers import l2
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import ReduceLROnPlateau, LearningRateScheduler
import tensorflow_datasets as tfds
import numpy as np
import matplotlib.pyplot as plt

def impHistoria(history):
    print(history.history.keys())
    plt.plot(history.history['accuracy']); plt.plot(history.history['val_accuracy'])
    plt.title('model accuracy'); plt.ylabel('accuracy'); plt.xlabel('epoch')
    plt.legend(['train', 'test'], loc='upper left'); plt.show()
    plt.plot(history.history['loss']); plt.plot(history.history['val_loss'])
    plt.title('model loss'); plt.ylabel('loss'); plt.xlabel('epoch')
    plt.legend(['train', 'test'], loc='upper left'); plt.show()

def load_stl10():
    ds_train, ds_test = tfds.load('stl10', split=['train', 'test'], as_supervised=True)

    ax = []
    ay = []
    for img, label in tfds.as_numpy(ds_train):
        ax.append(img)
        ay.append(label)

    qx = []
    qy = []
    for img, label in tfds.as_numpy(ds_test):
        qx.append(img)
        qy.append(label)

    ax = np.array(ax, dtype=np.float32)
    ay = np.array(ay, dtype=np.int32)
    qx = np.array(qx, dtype=np.float32)
    qy = np.array(qy, dtype=np.int32)

    return ax, ay, qx, qy

def resnet_layer(inputs, num_filters=16, kernel_size=3,
                 strides=1, activation='relu', batch_normalization=True):
    x = Conv2D(num_filters, kernel_size=kernel_size, strides=strides,
               padding='same', kernel_initializer='he_normal',
               kernel_regularizer=l2(1e-4))(inputs)
    if batch_normalization: x = BatchNormalization()(x)
    if activation is not None: x = Activation(activation)(x)
    return x

def lr_schedule(epoch):
    lr = 1e-3
    if epoch > 180: lr *= 0.5e-3
    elif epoch > 160: lr *= 1e-3
    elif epoch > 120: lr *= 1e-2
    elif epoch > 80: lr *= 1e-1
    print('Learning rate: ', lr)
    return lr

nomeprog="resnet_stl10_sem_aug"
batch_size = 32; num_classes = 10; epochs = 50
nl, nc = 96, 96

ax, ay, qx, qy = load_stl10()
input_shape = (nl, nc, 3)

ax = ax.astype('float32'); ax /= 255
qx = qx.astype('float32'); qx /= 255
ax -= 0.5; qx -= 0.5

ay = keras.utils.to_categorical(ay, num_classes)
qy = keras.utils.to_categorical(qy, num_classes)

inputs = Input(shape=input_shape)
x = resnet_layer(inputs=inputs)

num_filters = 16
y = resnet_layer(inputs=x, num_filters=num_filters, strides=1)
y = resnet_layer(inputs=y, num_filters=num_filters, activation=None)
x = keras.layers.add([x, y]); x = Activation('relu')(x)

y = resnet_layer(inputs=x, num_filters=num_filters, strides=1)
y = resnet_layer(inputs=y, num_filters=num_filters, activation=None)
x = keras.layers.add([x, y]); x = Activation('relu')(x)

y = resnet_layer(inputs=x, num_filters=num_filters, strides=1)
y = resnet_layer(inputs=y, num_filters=num_filters, activation=None)
x = keras.layers.add([x, y]); x = Activation('relu')(x)

num_filters *= 2
y = resnet_layer(inputs=x, num_filters=num_filters, strides=2)
y = resnet_layer(inputs=y, num_filters=num_filters, activation=None)
x = resnet_layer(inputs=x, num_filters=num_filters, kernel_size=1,
                 strides=2, activation=None, batch_normalization=False)
x = keras.layers.add([x, y]); x = Activation('relu')(x)

y = resnet_layer(inputs=x, num_filters=num_filters, strides=1)
y = resnet_layer(inputs=y, num_filters=num_filters, activation=None)
x = keras.layers.add([x, y]); x = Activation('relu')(x)

y = resnet_layer(inputs=x, num_filters=num_filters, strides=1)
y = resnet_layer(inputs=y, num_filters=num_filters, activation=None)
x = keras.layers.add([x, y]); x = Activation('relu')(x)

num_filters *= 2
y = resnet_layer(inputs=x, num_filters=num_filters, strides=2)
y = resnet_layer(inputs=y, num_filters=num_filters, activation=None)
x = resnet_layer(inputs=x, num_filters=num_filters, kernel_size=1,
                 strides=2, activation=None, batch_normalization=False)
x = keras.layers.add([x, y]); x = Activation('relu')(x)

y = resnet_layer(inputs=x, num_filters=num_filters, strides=1)
y = resnet_layer(inputs=y, num_filters=num_filters, activation=None)
x = keras.layers.add([x, y]); x = Activation('relu')(x)

y = resnet_layer(inputs=x, num_filters=num_filters, strides=1)
y = resnet_layer(inputs=y, num_filters=num_filters, activation=None)
x = keras.layers.add([x, y]); x = Activation('relu')(x)

x = GlobalAveragePooling2D()(x)
y = Flatten()(x)
outputs = Dense(num_classes, activation='softmax', kernel_initializer='he_normal')(y)

model = Model(inputs=inputs, outputs=outputs)
model.summary()

opt = Adam()
model.compile(optimizer=opt, loss='categorical_crossentropy', metrics=['accuracy'])

lr_scheduler = LearningRateScheduler(lr_schedule)
lr_reducer = ReduceLROnPlateau(factor=np.sqrt(0.1), cooldown=0, patience=5, min_lr=0.5e-6)
callbacks = [lr_reducer, lr_scheduler]

history = model.fit(ax, ay, batch_size=batch_size,
                    epochs=epochs, verbose=2, validation_data=(qx, qy),
                    callbacks=callbacks)

impHistoria(history)
score = model.evaluate(qx, qy, verbose=0)
print('Test loss: %.4f'%(score[0]))
print('Test accuracy: %.2f %%'%(100*score[1]))
print('Test error: %.2f %%'%(100*(1-score[1])))
model.save(nomeprog+'.keras')
```

```python
# resnet_stl10_prelayer_com_aug.py
# Rede inspirada em ResNet para classificar STL-10 com data augmentation

import os; os.environ['TF_CPP_MIN_LOG_LEVEL'] = '3'
os.environ['TF_FORCE_GPU_ALLOW_GROWTH'] = 'true'

import tensorflow.keras as keras
from tensorflow.keras.models import Model, load_model
from tensorflow.keras.layers import Dense, Conv2D, BatchNormalization, Activation
from tensorflow.keras.layers import GlobalAveragePooling2D, Input, Flatten
from tensorflow.keras.regularizers import l2
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import ReduceLROnPlateau, LearningRateScheduler
from tensorflow.keras.layers import RandomFlip, RandomRotation, RandomTranslation

import tensorflow_datasets as tfds
import numpy as np
import matplotlib.pyplot as plt


def impHistoria(history):
    print(history.history.keys())

    plt.figure()
    plt.plot(history.history['accuracy'])
    plt.plot(history.history['val_accuracy'])
    plt.title('model accuracy')
    plt.ylabel('accuracy')
    plt.xlabel('epoch')
    plt.legend(['train', 'test'], loc='upper left')
    plt.show()

    plt.figure()
    plt.plot(history.history['loss'])
    plt.plot(history.history['val_loss'])
    plt.title('model loss')
    plt.ylabel('loss')
    plt.xlabel('epoch')
    plt.legend(['train', 'test'], loc='upper left')
    plt.show()


def load_stl10():
    ds_train, ds_test = tfds.load(
        'stl10',
        split=['train', 'test'],
        as_supervised=True
    )

    ax = []
    ay = []
    for img, label in tfds.as_numpy(ds_train):
        ax.append(img)
        ay.append(label)

    qx = []
    qy = []
    for img, label in tfds.as_numpy(ds_test):
        qx.append(img)
        qy.append(label)

    ax = np.array(ax, dtype=np.float32)
    ay = np.array(ay, dtype=np.int32)
    qx = np.array(qx, dtype=np.float32)
    qy = np.array(qy, dtype=np.int32)

    return ax, ay, qx, qy


def resnet_layer(inputs, num_filters=16, kernel_size=3,
                 strides=1, activation='relu', batch_normalization=True):
    x = Conv2D(
        num_filters,
        kernel_size=kernel_size,
        strides=strides,
        padding='same',
        kernel_initializer='he_normal',
        kernel_regularizer=l2(1e-4)
    )(inputs)

    if batch_normalization:
        x = BatchNormalization()(x)

    if activation is not None:
        x = Activation(activation)(x)

    return x


def lr_schedule(epoch):
    lr = 1e-3
    if epoch > 180:
        lr *= 0.5e-3
    elif epoch > 160:
        lr *= 1e-3
    elif epoch > 120:
        lr *= 1e-2
    elif epoch > 80:
        lr *= 1e-1
    print('Learning rate: ', lr)
    return lr


nomeprog = "resnet_stl10_com_aug"

batch_size = 32
num_classes = 10
epochs = 50

nl, nc = 96, 96

# ------------------------------------------------------------
# STL-10 no lugar de CIFAR-10
# ------------------------------------------------------------
ax, ay, qx, qy = load_stl10()
input_shape = (nl, nc, 3)

print("Shape ax:", ax.shape)
print("Shape ay:", ay.shape)
print("Shape qx:", qx.shape)
print("Shape qy:", qy.shape)

# ------------------------------------------------------------
# Normalização
# ------------------------------------------------------------
ax = ax.astype('float32'); ax /= 255.0
qx = qx.astype('float32'); qx /= 255.0

# igual ao script original: intervalo [-0.5, +0.5]
ax -= 0.5
qx -= 0.5

ay = keras.utils.to_categorical(ay, num_classes)
qy = keras.utils.to_categorical(qy, num_classes)

# ------------------------------------------------------------
# Modelo com data augmentation nas preprocessing layers
# ------------------------------------------------------------
inputs = Input(shape=input_shape)

x = RandomRotation(0.042, fill_mode="nearest", interpolation="bilinear")(inputs)
x = RandomTranslation(0.1, 0.1, fill_mode="nearest", interpolation="bilinear")(x)
x = RandomFlip("horizontal")(x)

x = resnet_layer(inputs=x)

num_filters = 16

# ---------------- bloco residual 1 ----------------
y = resnet_layer(inputs=x, num_filters=num_filters, strides=1)
y = resnet_layer(inputs=y, num_filters=num_filters, activation=None)
x = keras.layers.add([x, y]); x = Activation('relu')(x)

y = resnet_layer(inputs=x, num_filters=num_filters, strides=1)
y = resnet_layer(inputs=y, num_filters=num_filters, activation=None)
x = keras.layers.add([x, y]); x = Activation('relu')(x)

y = resnet_layer(inputs=x, num_filters=num_filters, strides=1)
y = resnet_layer(inputs=y, num_filters=num_filters, activation=None)
x = keras.layers.add([x, y]); x = Activation('relu')(x)

# ---------------- bloco residual 2 ----------------
num_filters *= 2

y = resnet_layer(inputs=x, num_filters=num_filters, strides=2)
y = resnet_layer(inputs=y, num_filters=num_filters, activation=None)

x = resnet_layer(inputs=x, num_filters=num_filters, kernel_size=1,
                 strides=2, activation=None, batch_normalization=False)
x = keras.layers.add([x, y]); x = Activation('relu')(x)

y = resnet_layer(inputs=x, num_filters=num_filters, strides=1)
y = resnet_layer(inputs=y, num_filters=num_filters, activation=None)
x = keras.layers.add([x, y]); x = Activation('relu')(x)

y = resnet_layer(inputs=x, num_filters=num_filters, strides=1)
y = resnet_layer(inputs=y, num_filters=num_filters, activation=None)
x = keras.layers.add([x, y]); x = Activation('relu')(x)

# ---------------- bloco residual 3 ----------------
num_filters *= 2

y = resnet_layer(inputs=x, num_filters=num_filters, strides=2)
y = resnet_layer(inputs=y, num_filters=num_filters, activation=None)

x = resnet_layer(inputs=x, num_filters=num_filters, kernel_size=1,
                 strides=2, activation=None, batch_normalization=False)
x = keras.layers.add([x, y]); x = Activation('relu')(x)

y = resnet_layer(inputs=x, num_filters=num_filters, strides=1)
y = resnet_layer(inputs=y, num_filters=num_filters, activation=None)
x = keras.layers.add([x, y]); x = Activation('relu')(x)

y = resnet_layer(inputs=x, num_filters=num_filters, strides=1)
y = resnet_layer(inputs=y, num_filters=num_filters, activation=None)
x = keras.layers.add([x, y]); x = Activation('relu')(x)

x = GlobalAveragePooling2D()(x)
y = Flatten()(x)
outputs = Dense(num_classes, activation='softmax', kernel_initializer='he_normal')(y)

# ------------------------------------------------------------
# Construção do modelo
# ------------------------------------------------------------
model = Model(inputs=inputs, outputs=outputs)
# model = load_model(nomeprog + '.keras')
model.summary()

opt = Adam()
model.compile(optimizer=opt, loss='categorical_crossentropy', metrics=['accuracy'])

lr_scheduler = LearningRateScheduler(lr_schedule)
lr_reducer = ReduceLROnPlateau(
    factor=np.sqrt(0.1),
    cooldown=0,
    patience=5,
    min_lr=0.5e-6
)
callbacks = [lr_reducer, lr_scheduler]

history = model.fit(
    ax, ay,
    batch_size=batch_size,
    epochs=epochs,
    verbose=2,
    validation_data=(qx, qy),
    callbacks=callbacks
)

impHistoria(history)

score = model.evaluate(qx, qy, verbose=0)
print('Test loss: %.4f' % (score[0]))
print('Test accuracy: %.2f %%' % (100 * score[1]))
print('Test error: %.2f %%' % (100 * (1 - score[1])))

model.save(nomeprog + '.keras')
```

```python
# resnet_stl10_tta.py
# Usa o modelo treinado da lição de casa #1 para classificar STL-10 com TTA de 11 distorções

import os; os.environ['TF_CPP_MIN_LOG_LEVEL'] = '3'
os.environ['TF_FORCE_GPU_ALLOW_GROWTH'] = 'true'

import tensorflow.keras as keras
from tensorflow.keras.models import load_model
import tensorflow_datasets as tfds
import tensorflow as tf
import numpy as np


def load_stl10_test():
    ds_test = tfds.load('stl10', split='test', as_supervised=True)

    qx = []
    qy = []
    for img, label in tfds.as_numpy(ds_test):
        qx.append(img)
        qy.append(label)

    qx = np.array(qx, dtype=np.float32)
    qy = np.array(qy, dtype=np.int32)

    return qx, qy


def shift_img(img, dx, dy):
    """
    Desloca a imagem em pixels.
    dx > 0 desloca para a direita
    dy > 0 desloca para baixo
    """
    return tf.roll(img, shift=[dy, dx], axis=[0, 1])


def rotate_img(img, angle_rad):
    """
    Rotação usando tensorflow raw op.
    """
    img4 = tf.expand_dims(img, 0)
    out = tf.image.rot90(img4, 0)  # só para garantir tensor 4D
    # usa affine via keras preprocessing seria mais simples, mas aqui deixamos em tf puro
    # como TTA simples, podemos usar projetoive transform via raw ops:
    c = tf.math.cos(angle_rad)
    s = tf.math.sin(angle_rad)

    h = tf.cast(tf.shape(img)[0], tf.float32)
    w = tf.cast(tf.shape(img)[1], tf.float32)
    cx = (w - 1.0) / 2.0
    cy = (h - 1.0) / 2.0

    # transformação afim ao redor do centro
    a0 = c
    a1 = -s
    a2 = cx - c * cx + s * cy
    a3 = s
    a4 = c
    a5 = cy - s * cx - c * cy
    a6 = 0.0
    a7 = 0.0

    transforms = tf.convert_to_tensor([[a0, a1, a2, a3, a4, a5, a6, a7]], dtype=tf.float32)

    out = tf.raw_ops.ImageProjectiveTransformV3(
        images=img4,
        transforms=transforms,
        output_shape=tf.shape(img)[0:2],
        interpolation="BILINEAR",
        fill_value=0.0
    )
    return out[0]


def zoom_in(img, frac=0.9):
    """
    Zoom in: recorta o centro e redimensiona de volta.
    frac menor que 1 recorta uma região central.
    """
    h = tf.shape(img)[0]
    w = tf.shape(img)[1]

    nh = tf.cast(tf.cast(h, tf.float32) * frac, tf.int32)
    nw = tf.cast(tf.cast(w, tf.float32) * frac, tf.int32)

    y0 = (h - nh) // 2
    x0 = (w - nw) // 2

    crop = img[y0:y0+nh, x0:x0+nw, :]
    crop = tf.image.resize(crop, [h, w], method='bilinear')
    return crop


def zoom_out(img, frac=0.9):
    """
    Zoom out: reduz a imagem e centraliza num canvas do tamanho original.
    """
    h = tf.shape(img)[0]
    w = tf.shape(img)[1]

    nh = tf.cast(tf.cast(h, tf.float32) * frac, tf.int32)
    nw = tf.cast(tf.cast(w, tf.float32) * frac, tf.int32)

    small = tf.image.resize(img, [nh, nw], method='bilinear')

    pad_top = (h - nh) // 2
    pad_bottom = h - nh - pad_top
    pad_left = (w - nw) // 2
    pad_right = w - nw - pad_left

    out = tf.pad(small, [[pad_top, pad_bottom], [pad_left, pad_right], [0, 0]])
    return out


def build_11_tta_versions(img):
    """
    Gera 11 versões da imagem:
    1) original
    2) shift direita
    3) shift esquerda
    4) shift baixo
    5) shift cima
    6) rotação +5 graus
    7) rotação -5 graus
    8) zoom in
    9) zoom out
    10) flip horizontal
    11) rotação + translação leve
    """
    deg = np.pi / 180.0

    imgs = []
    imgs.append(img)                              # 1
    imgs.append(shift_img(img,  3,  0))         # 2
    imgs.append(shift_img(img, -3,  0))         # 3
    imgs.append(shift_img(img,  0,  3))         # 4
    imgs.append(shift_img(img,  0, -3))         # 5
    imgs.append(rotate_img(img,  5 * deg))      # 6
    imgs.append(rotate_img(img, -5 * deg))      # 7
    imgs.append(zoom_in(img, 0.9))              # 8
    imgs.append(zoom_out(img, 0.9))             # 9
    imgs.append(tf.image.flip_left_right(img))  # 10

    img11 = rotate_img(img, 3 * deg)
    img11 = shift_img(img11, 2, -2)
    imgs.append(img11)                          # 11

    return tf.stack(imgs, axis=0)


def predict_tta(model, qx, batch_size=64):
    """
    Para cada imagem, gera 11 distorções, faz predição,
    tira a média das probabilidades e escolhe a classe final.
    """
    preds = []

    for i in range(len(qx)):
        img = tf.convert_to_tensor(qx[i], dtype=tf.float32)
        batch_11 = build_11_tta_versions(img).numpy()   # (11, 96, 96, 3)

        # mesmo pré-processamento do treino: [0,1] -> [-0.5, 0.5]
        batch_11 = batch_11 / 255.0
        batch_11 = batch_11 - 0.5

        out = model.predict(batch_11, verbose=0)        # (11, 10), já em softmax
        mean_out = np.mean(out, axis=0)                 # média das 11 probabilidades
        pred = np.argmax(mean_out)

        preds.append(pred)

        if i % 500 == 0:
            print(f"{i}/{len(qx)}")

    return np.array(preds, dtype=np.int32)


nomeprog = "resnet_stl10_com_aug.keras"

# ------------------------------------------------------------
# carrega modelo treinado na lição #1
# ------------------------------------------------------------
model = load_model(nomeprog)
model.summary()

# ------------------------------------------------------------
# carrega teste STL-10
# ------------------------------------------------------------
qx, qy = load_stl10_test()
print("Shape qx:", qx.shape)
print("Shape qy:", qy.shape)

# ------------------------------------------------------------
# predição com TTA
# ------------------------------------------------------------
pred = predict_tta(model, qx)

acc = np.mean(pred == qy)
err = 1.0 - acc

print("Test accuracy with TTA: %.2f %%" % (100 * acc))
print("Test error with TTA: %.2f %%" % (100 * err))
print("Acuracidade com TTA: %.2f %%" % (100 * acc))
```
